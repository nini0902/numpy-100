#### 27. Consider an integer vector Z, which of these expressions are legal? (★☆☆)
```python
Z**Z
2 << Z >> 2
Z <- Z
1j*Z
Z/1/1
Z<Z>Z
```

````python
Z = np.arange(5)
print(Z**Z) # element-wise exponentiation
print(2 << Z >> 2) # element-wise bitwise shift
print(Z < -Z) # element-wise comparison(check is each element of Z is less than its negation)
print(1j * Z) # element-wise multiplication with imaginary unit
print(Z / 1 / 1) # element-wise division, the result is a float array, not an integer array
print(Z < Z > Z) # element-wise comparison (illegal)
````

#### 29. How to round away from zero a float array ? (★☆☆)

````python
Z = np.array([-1.7, -1.5, -0.2, 0.1, 0.6, 1.9, 2.5, 3.3])
# to round away from zero, ignore the sign first, round up, then put the sign back
result = np.copysign(np.ceil(np.abs(Z)), Z)
print(result)
````

#### 33. How to get the dates of yesterday, today and tomorrow? (★☆☆)

````python
yesterday = np.datetime64("today") - np.timedelta64(1, "D")
today = np.datetime64("today")
tomorrow = np.datetime64("today") + np.timedelta64(1, "D")

print("yesterday: ", yesterday)
print("today: ", today)
print("tomorrow: ", tomorrow)
````

#### 35. How to compute ((A+B)*(-A/2)) in place (without copy)? (★★☆)

````python
A = np.array([1., 2., 3.])
B = np.array([4., 5., 6.])

np.add(A,B,out = B)
np.divide(A,2, out = A)
np.negative(A, out = A)
np.multiply(A,B, out = A)

print(A)
````

#### 38. Consider a generator function that generates 10 integers and use it to build an array (★☆☆)
